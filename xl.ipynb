```python

from openpyxl import load_workbook
from openpyxl.styles import PatternFill
src = load_workbook(r"C:\Users\anuma\Downloads\employee_source.xlsx").active
dest_wb = load_workbook(r"C:\Users\anuma\Downloads\employee_destination.xlsx")
dest = dest_wb.active
red = PatternFill(start_color="FF0000", end_color="FF0000", fill_type="solid")
for row in src.iter_rows(min_row=2, values_only=True):
    dest.append(row)
    if row[1] > 40000:
        for cell in dest[dest.max_row]:
            cell.fill = red
dest_wb.save(r"C:\Users\anuma\Downloads\employee_destination.xlsx")
print("Data copied and high salaries highlighted!")
```

```python

from openpyxl import load_workbook
from openpyxl.styles import PatternFill

src = load_workbook(r"C:\Users\anuma\Downloads\employee_destination.xlsx").active
wb = load_workbook(r"C:\Users\anuma\Downloads\highlighted_employees_destination.xlsx")
dest = wb.active

for row in src.iter_rows(min_row=2):
    if row[0].fill.fgColor.rgb in ("FFFF0000", "00FF0000"):
        dest.append([cell.value for cell in row])

wb.save(r"C:\Users\anuma\Downloads\highlighted_employees_destination.xlsx")
print("Red-highlighted employees copied!")
```
